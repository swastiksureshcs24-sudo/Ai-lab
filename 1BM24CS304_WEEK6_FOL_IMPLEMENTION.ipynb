```python

def is_variable(term):
    """Variables are lowercase strings (e.g., 'x', 'y')."""
    return isinstance(term, str) and term[0].islower()

def is_constant(term):
    """Constants are uppercase/capitalized strings (e.g., 'John', 'A')."""
    return isinstance(term, str) and not term[0].islower()

def occurs_in(var, expr):
    """Checks if variable 'var' occurs anywhere inside 'expr'."""
    if var == expr:
        return True
    if isinstance(expr, list):
        return any(occurs_in(var, sub_expr) for sub_expr in expr)
    return False

def apply_substitution(subst, expr):
    """Applies a list of substitutions [(val, var), ...] to an expression."""
    if not subst:
        return expr
    for val, var in subst:
        expr = substitute_single(val, var, expr)
    return expr

def substitute_single(val, var, expr):
    """Replaces all occurrences of 'var' with 'val' in 'expr'."""
    if isinstance(expr, str):
        return val if expr == var else expr
    if isinstance(expr, list):
        return [substitute_single(val, var, item) for item in expr]
    return expr

def unify(psi1, psi2):
    """
    Unify(Ψ1, Ψ2) following the exact steps from the algorithm:
    - Compound expressions (predicates/functions) are lists: ['P', 'x', ['f', 'y']]
    - Returns [] for NIL, "FAILURE" on failure, or a list of (replacement, variable) tuples.
    """

    if isinstance(psi1, str) or isinstance(psi2, str):

        if psi1 == psi2:
            return []


        elif is_variable(psi1):
            if occurs_in(psi1, psi2):
                return "FAILURE"
            else:
                return [(psi2, psi1)]


        elif is_variable(psi2):
            if occurs_in(psi2, psi1):
                return "FAILURE"
            else:
                return [(psi1, psi2)]

        else:
            return "FAILURE"

    if psi1[0] != psi2[0]:
        return "FAILURE"


    if len(psi1) != len(psi2):
        return "FAILURE"


    subst = []


    l1 = list(psi1)
    l2 = list(psi2)

    for i in range(1, len(l1)):

        s = unify(l1[i], l2[i])


        if s == "FAILURE":
            return "FAILURE"


        if s != []:
            for j in range(i + 1, len(l1)):
                l1[j] = apply_substitution(s, l1[j])
                l2[j] = apply_substitution(s, l2[j])
            subst.extend(s)

    return subst

def format_output(result):
    """Formats the substitution set to match textbook notation {(val/var)}."""
    if result == "FAILURE":
        return "FAILURE"
    if not result:
        return "NIL"
    pairs = [f"{val}/{var}" for val, var in result]
    return "{ " + ", ".join(pairs) + " }"



if __name__ == "__main__":

    expr1 = ['P', 'x', ['f', 'y']]
    expr2 = ['P', 'John', ['f', 'z']]
    print("Test 1:", format_output(unify(expr1, expr2)))



    expr3 = ['P', 'x', 'x']
    expr4 = ['P', 'y', ['f', 'y']]
    print("Test 2:", format_output(unify(expr3, expr4)))



```
