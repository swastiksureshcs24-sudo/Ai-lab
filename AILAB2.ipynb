```python
def dfs(start, goal):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path + [state]

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        zero = state.index(0)
        row, col = zero // 3, zero % 3

        moves = [
            (-1, 0),  # Up
            (1, 0),   # Down
            (0, -1),  # Left
            (0, 1)    # Right
        ]

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new_pos = r * 3 + c
                new_state = state.copy()

                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )

                if tuple(new_state) not in visited:
                    stack.append((new_state, path + [state]))

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

result = dfs(start, goal)

if result:
    print("Solution found:")
    for state in result:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
else:
    print("No solution found")
```

```python
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            new_state = list(state)
            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )
            neighbors.append(tuple(new_state))

    return neighbors


def dls(state, depth, path):
    if state == goal:
        return path

    if depth == 0:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in path:
            result = dls(neighbor, depth - 1, path + [neighbor])

            if result is not None:
                return result

    return None


def ids(start, max_depth):
    for limit in range(max_depth + 1):
        result = dls(start, limit, [start])

        if result is not None:
            print("\nSolution found at depth:", limit)
            return result

    return None


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


# Get depth limit from the user
max_depth = int(input("Enter maximum depth limit: "))

solution = ids(start, max_depth)

if solution:
    for i, state in enumerate(solution):
        print("Step", i)
        print_puzzle(state)
else:
    print("No solution found within the given depth limit.")
```
